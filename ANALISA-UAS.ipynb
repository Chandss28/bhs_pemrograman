## Jawabannya
Mari kita jabarkan menggunakan pendekatan 5W 1H (What, Who, Where, When, Why, How):

1. **What (Apa)**: Data pelanggan adalah informasi tentang individu atau entitas yang menggunakan produk atau layanan dari sebuah perusahaan.
   
2. **Who (Siapa)**: Individu atau entitas yang menjadi subjek data pelanggan adalah mereka yang aktif menggunakan produk atau layanan perusahaan tertentu.

3. **Where (Dimana)**: Informasi ini berhubungan dengan pelanggan yang menggunakan produk atau layanan di tempat yang relevan dengan aktivitas bisnis perusahaan tersebut.

4. **When (Kapan)**: Data pelanggan mencakup riwayat transaksi, yang mencatat kapan pelanggan melakukan interaksi atau pembelian dengan perusahaan.

5. **Why (Mengapa)**: Tujuan dari pengumpulan data pelanggan adalah untuk membantu perusahaan memahami kebutuhan, preferensi, dan perilaku pelanggan untuk meningkatkan layanan dan kepuasan pelanggan.

6. **How (Bagaimana)**: Data pelanggan dikumpulkan melalui berbagai cara, seperti formulir pendaftaran, transaksi pembelian, survei kepuasan pelanggan, dan interaksi langsung dengan layanan pelanggan.

Dengan menggunakan pendekatan 5W 1H ini, kita dapat memahami dengan lebih baik apa itu data pelanggan dan bagaimana informasi tersebut digunakan oleh perusahaan untuk kepentingan mereka dan juga kepuasan pelanggan.
